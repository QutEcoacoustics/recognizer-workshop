# Panel App for Remote Deployment

This notebook sets up the Recognizer Workshop for deployment via `panel serve`.

**To deploy on a remote server:**

1. Run the first cell to set up the app
2. In terminal, run: 
   ```bash
   panel serve app.ipynb --allow-websocket-origin=* --port=5006 --address=0.0.0.0
   ```
3. Access via: `http://your-server-address:5006/app`

The `--address=0.0.0.0` is crucial for remote access - without it, Panel only binds to localhost.
